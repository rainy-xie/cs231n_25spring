## Inline Question 1:

We've only asked you to implement ReLU, but there are a number of different activation functions that one could use in neural networks, each with its pros and cons. In particular, an issue commonly seen with activation functions is getting zero (or close to zero) gradient flow during backpropagation. Which of the following activation functions have this problem? If you consider these functions in the one dimensional case, what types of input would lead to this behaviour?
1. Sigmoid
2. ReLU
3. Leaky ReLU

$\color{blue}{\textit Your Answer:}$

1,2


Sigmoid：✓ 有梯度接近零的问题
Sigmoid函数公式：σ(x) = 1/(1 + e^(-x))
导数：σ'(x) = σ(x)(1 - σ(x))
当输入x的绝对值很大时，梯度会趋近于零：

当x >> 0（非常大的正数）时，σ(x)接近1，导致σ'(x)接近0

当x << 0（非常大的负数）时，σ(x)接近0，导致σ'(x)也接近0

这种现象称为"梯度消失"，会导致深层网络训练困难。

ReLU：✓ 有梯度为零的问题
ReLU函数公式：f(x) = max(0, x)
导数：f'(x) = 1, 当x > 0; f'(x) = 0, 当x ≤ 0
当输入为负值时，ReLU的梯度完全为零，这会导致"死亡ReLU"问题 —— 一旦神经元进入负值区域，它可能永远不会被激活，因为梯度为零使得权重无法通过反向传播更新。

Leaky ReLU：𝗫 理论上不存在梯度完全为零的问题

Leaky ReLU函数公式：f(x) = αx, 当x < 0; f(x) = x, 当x ≥ 0（α是一个小正数，通常为0.01）
导数：f'(x) = α, 当x < 0; f'(x) = 1, 当x ≥ 0

Leaky ReLU正是为了解决ReLU的"死亡"问题而设计的，它在负值区域提供了一个小的非零梯度α，确保即使在负值区域也能进行权重更新。虽然在负值区域梯度很小，但严格来说不会为零。

Train on a larger dataset:

增加训练数据可以减少过拟合，因为模型会从更多的数据中学习到更通用的特征，而不是记住训练集中的特定模式。这通常是减少训练集和测试集准确率差距的最有效方法。

Add more hidden units:

增加隐藏单元会增加模型的容量，但这可能会导致过拟合，从而加大训练集和测试集之间的差距。因此，这不是减少差距的有效方法。

Increase the regularization strength:

增加正则化强度（如 L2 正则化）可以限制模型的复杂度，防止模型过拟合训练数据，从而减少训练集和测试集之间的准确率差距。
